```python
import torch
import numpy as np
```

```python
a = torch.tensor([[1, 2], [3, 4]])
print(a)
```

```python
b = np.array([[1, 2], [3, 4]])
print(b)
```

```python
c = torch.from_numpy(b)
print(c)
```

```python
d = c.numpy()
print(d)
```

```python
e = torch.ones((2, 3))
print(e)
```

```python
f = torch.zeros((2, 3))
print(f)
```

```python
g = torch.rand((2, 3))
print(g)
```

```python
if torch.cuda.is_available():
    device = torch.device("cuda")
    print("CUDA is available. Using GPU.")
else:
    device = torch.device("cpu")
    print("CUDA is not available. Using CPU.")
```
